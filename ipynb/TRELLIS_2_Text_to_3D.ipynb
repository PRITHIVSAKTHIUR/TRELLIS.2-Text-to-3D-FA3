## **TRELLIS.2-Text-to-3D**
TRELLIS.2-Text-to-3D is an end-to-end Text-to-3D and Image-to-3D generation app that enables users to create high-quality 3D GLB assets either by generating an image from a text prompt or by uploading an existing image, powered by Z-Image-Turbo and the TRELLIS.2 multi-stage pipeline. It runs fully on GPU with automatic background removal and smart image preprocessing, supports advanced controls for structure, shape, material sampling, resolution, mesh decimation, and texture size, and provides an interactive 3D preview with one-click GLB download, making it ideal for rapid ideation and production-ready 3D asset creation.
